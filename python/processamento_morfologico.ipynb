```python
import numpy as np
```

```python
def create_mask(size=3, form="square"):
    if size % 2 == 0:
        size += 1
    
    kernel = np.zeros((size, size), dtype=np.uint8)
    center = size // 2

    if form == "square":
        kernel = np.ones((size, size), dtype=np.uint8)

    elif form == "cross":
        kernel[center, :] = 1
        kernel[:, center] = 1
    
    elif form == "circle":
        for i in range(size):
            for j in range(size):
                if ((i - center)**2 + (j - center)**2 <= center**2):
                    kernel[i,j] = 1
    
    return kernel
```
